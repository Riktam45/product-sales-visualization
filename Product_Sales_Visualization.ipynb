```python
import matplotlib.pyplot as plt
#sample temp data for different arrays

days = [1,2,3,4,5]
temperature = [20,22,20,22,21]

#plot the data
plt.plot(days, temperature, marker = 'o', linestyle = '-', color = 'b', label = 'temperature')

#adding lable & title
plt.xlabel('Days')
plt.ylabel('Temperature( in c)')
plt.title('Temperature variation over days')

#Display the plot
plt.legend()
plt.grid(True)
plt.show()
```

```python
# using seaborn

import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    'category':['Product A','Product B','Product C','Product D'],'value':[150,200,250,300]
}
df = pd.DataFrame(data)

#create a barchart using barplot()
sns.barplot(x ='category',y = 'value', data = df)

#customized the plot
sns.set_style('whitegrid')
sns.set_color_codes('pastel')

#adding the lable and titles
sns.xlabel('Product category')
sns.ylabel('Purchase value')
sns.title('Customer purchase by producer category')

plt.show()
```
